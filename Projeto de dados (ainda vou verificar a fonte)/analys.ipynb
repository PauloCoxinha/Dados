```python
import pandas as pd 

from push import df_bronze
```

```python
df_silver = df_bronze.copy()

colunas_de_preco = ["preco_menor", "preco_comum", "preco_maior"]

for coluna in colunas_de_preco:
    df_silver[coluna] = df_silver[coluna].str.replace(r'[^\d,]', '', regex=True)

    df_silver[coluna] = df_silver[coluna].str.replace(',', '.').astype(float)

df_silver['Produto'] = df_silver['Produto'].str.upper()

print("\n--- Tipos de dados na camada Silver ---")
print(df_silver.dtypes)

print("\n--- Amostra dos dados limpos ---")
print(df_silver.head())
```
